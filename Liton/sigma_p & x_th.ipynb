### 模型层（lev）的含义

- Eckermann, S. (2009). Hybrid σ–p Coordinate Choices for a Global Model. Monthly Weather Review, 137(1), 224-245. https://doi.org/10.1175/2008MWR2537.1

- 孙敬哲.基于气压的地形跟随混合坐标算法优化及应用[D].国防科学技术大学[2025-11-06].DOI:CNKI:CDMD:2.1017.834889.

#### 模型采用的是   $\sigma -p$混合坐标

- 广义垂直坐标系：
$$
\eta = h(P,P_s)
$$
归一化取为：
$$
h(P_s,P_s) = 1 \\
h(P_{top},P_s) = 0
$$
$P_s$为地表附近气压，$h(p,p_s)$为$p$的单调函数，对于给定的地面气压 $p_s$，气压 $p$ 和 $\eta$ 之间形成一一映射关系

1. $p$坐标：一个以气压p为垂直坐标变量的垂直坐标系。每个坐标对应一个等压面。然而，当在模式中考虑地形变化后，$p$坐标系在地形附近遇到计算困难，气压坐标系下的大气下边界不与坐标面重合，因此等压面与地形相交会导致出现“空洞”，这对模式的正常运行造成了不良影响。

2. $\sigma$坐标（数值天气预报广泛使用的坐标系）：定义为取广义垂直坐标为$\sigma$
$$
\eta = h(P,P_s) = \frac{P-P_{top}}{P_s-P_{top}}=\sigma
$$
$P_{top}$一般取某一常数或取为零（`MERRA-2`中取为 $0.01hPa$），$P_s=P_s(x,y,t)$是地表面气压。

在$\sigma$坐标系下，大气上界与坐标面（$\sigma=0$）为了一个坐标面重合，地球表面也成为了一个坐标面（$\sigma=1$），使得模式的下边界变得十分简单。

3. 两者综合就是`混合坐标系`：在低层采用$\sigma$坐标系，所以能够表示地形的变化；但是到了高层之后，随高度升高坐标面对地形变化的响应逐渐减小，最终过渡高层的$p$坐标系，$h(p,p_s)$形式采用为:
$$
p(\tilde{\eta},p_s)=A(\tilde{\eta})+B(\tilde{\eta})(p_s-p_{top})
$$
其中 $\tilde{\eta}$ 为 $p_s=p_0$ 时的 $\eta=h(p,p_0)$ 垂直轮廓线，$p_0 \approx 1000hPa$ 为海平面参考气压。细节就不再赘述。
$$
\eta = \frac{p-p_{top}}{p_s-p_{top}}=\frac{A(\tilde{\eta})-p_{top}}{p_s-p_{top}}+B(\tilde{\eta})
$$

回到模型层本身，最终选择使用海平面参考气压 $p_0=1000\ hPa$ 下对应的 $\tilde{\eta}$ 值标记模式层。总层数为$L$，在离散形式下，模式整层(Layer)自顶到底标记为：
$$
k=1,2,3,4,\dots,L
$$
模式半层(Interface)自顶到底标记为:
$$
k=\tilde{0},\tilde{1},,\cdots,\tilde{L}
$$
模式半层就是相邻两个模式层的分界线。半层上的气压由下式给定：
$$
p_{k+1/2}=A_{k+1/2}+B_{k+1/2}(p_s-p_{top})
$$

我们一直在讨论的模型层lev，其每一层对应一个固定的$\tilde{\eta}$值，$\tilde{\eta}$来自$A_{k+1/2}$和$B_{k-1/2}$，通过后两者以及上式算出$p_{k+1/2}$，然后用$\eta$计算公式和参数$p_s=p_0$算出$\tilde{\eta}$

```python
# ['H', 'DELP', 'PL', 'QV', 'T']
# T: 温度(K), DELP: 气压厚度(Pa), QV: 比湿(kg/kg), PS: 地表气压(Pa)

# More efficient memory usage
def select_variables(ds):
    """Only keep variables that we need"""
    variables_to_keep = ['H', 'DELP', 'PL', 'QV', 'T']
    available_vars = [var for var in variables_to_keep if var in ds.data_vars]
    return ds[available_vars]
```
